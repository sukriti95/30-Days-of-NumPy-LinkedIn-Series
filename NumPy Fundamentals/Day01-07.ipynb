### Day 01 – Introduction to NumPy

#### NumPy (Numerical Python) is a Python library designed for efficient numerical computations and working with multidimensional arrays.


### 1. Using Python list

```python
sales = [1000,2000,3000,4000]
```

```python
updated_sales = [x * 1.10 for x in sales]
```

```python
print(updated_sales)
```

### 2.Using NumPy

```python
import numpy as np
```

```python
sales = np.array([1000,2000,3000,4000])
```

```python
updated_sales = sales * 1.10
```

```python
print(updated_sales)
```

### Day 02 – Creating NumPy Arrays

### 1. Creating an array from a Python list

```python
sales = [12000, 15000, 18000, 22000]
```

```python
sales_array = np.array(sales)
```

```python
print(sales_array)
```

### 2. Creating a 2D array

```python
monthly_sales = np.array([
    [12000, 15000],
    [18000, 22000],
    [14000, 16000],
    [20000, 25000]
])

print(monthly_sales)
```

```python
print(monthly_sales)
```

### 3. Creating arrays using built-in functions

```python
# Array filled with zeros
np.zeros((2, 3))
```

```python
# Array filled with ones
np.ones((2, 3))
```

```python
# Array with a fixed value
np.full((2, 3), 100)
```

```python
# Generate numbers within a range
np.arange(1, 11, 2)
```

```python
# Generate evenly spaced values
np.linspace(0, 100, 5)
```

### 4. Inspecting the array

```python
print(monthly_sales.ndim)     # ndim: Number of dimensions.
```

```python
print(monthly_sales.shape)    # shape: Number of rows and columns.
```

```python
print(monthly_sales.size)     # size: Total number of elements.
```

```python
print(monthly_sales.dtype)   #dtype: Data type of the elements.
```

 # Day 03 – Array Dimensions and Shape


#### 1. One-Dimensional Array (1D)

```python
daily_sales = np.array([1200, 1500, 1800, 2000, 2500])
```

```python
print(daily_sales.ndim)
```

```python
print(daily_sales.shape)
```

#### 2. Two-Dimensional Array (2D)


```python
regional_sales = np.array([
    [12000, 15000, 18000, 20000],
    [14000, 16000, 19000, 22000],
    [10000, 13000, 17000, 21000]
])
```

```python
print(regional_sales.ndim)
```

```python
print(regional_sales.shape)
```

#### 3. Three-Dimensional Array (3D)

```python
yearly_sales = np.array([
    [
        [12000, 15000, 18000, 20000],
        [14000, 16000, 19000, 22000],
        [10000, 13000, 17000, 21000]
    ],
    [
        [15000, 17000, 20000, 23000],
        [16000, 19000, 21000, 25000],
        [12000, 15000, 19000, 24000]
    ]
])
```

```python
print(yearly_sales.ndim)
```

```python
print(yearly_sales.shape)
```
